## Get The Model

First, we will get our registered model from MLFlow.

```python
import dagshub
dagshub.init(repo_owner='zhorzholianimate', repo_name='House-Prices-Advanced-Regression-Techniques', mlflow=True)

import mlflow
```

```python
model_name = "outliers_remover_model"
model_version = '1'  # Use the correct version here

model_uri = f"models:/{model_name}/{model_version}"
model = mlflow.pyfunc.load_model(model_uri)
```

## Predict 

Import some libraries.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Get X_test dataset first.

```python
X_test = pd.read_csv('./data/test.csv')
X_test.shape
```

Make Prediction.

```python
predictions = model.predict(X_test)
predictions
```

Make predictions in sumbision form.

```python
predictions_df = pd.DataFrame({
    'Id': X_test['Id'], 
    'SalePrice': predictions
})
predictions_df.head()
```

Save the file.

```python
predictions_df.to_csv('answer.csv', index=False)
```
